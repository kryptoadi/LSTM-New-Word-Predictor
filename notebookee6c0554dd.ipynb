```python
# import pandas as pd

# path = "/content/drive/MyDrive/nextwordpredict/News_Category_Dataset_v3.json"

# df = pd.read_json(path, lines=True)

# Take 40% of the dataset

# import pandas as pd
# df=pd.read_json('News_Category_Dataset_v3.json',lines=True)
# df = df.sample(frac=0.40, random_state=42)

# print(df.shape)
import pandas as pd
df = pd.read_json("/kaggle/input/datasets/adityaraj30/news-category-dataset-v3/News_Category_Dataset_v3.json",
                  lines=True)
df = df.sample(frac=0.30, random_state=42)
```

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,LSTM,Dense
```

```python
tokenizer=Tokenizer(oov_token='<OOV>')
```

```python
tokenizer.fit_on_texts(df)
```

```python
len(tokenizer.index_word)
```

```python
tokenizer.word_counts
```

```python
df=tokenizer.texts_to_sequences(df)
```

```python
df
```

```python
input_sequences=[]
for seq in df:
    if len(seq)<2:
        continue
    for i in range(1,len(seq)):
        ngram=seq[:i+1]
        input_sequences.append(ngram)
```

```python
max_len=max(len(i) for i in input_sequences)
```

```python
max_len
```

```python
input=pad_sequences(input_sequences,padding='pre',maxlen=max_len)
```

```python
input.shape
```

```python
x=input[:,:-1]
y=input[:,-1]
```

```python
y.shape
```

```python
x.shape
```

```python
y=np.array(y)
```

```python
x[0]
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train.shape
```

```python
y_train.shape
```

```python
vocab_size=len(tokenizer.word_index)+1
```

```python
vocab_size
```

```python
max_len=x.shape[1]
```

```python
max_len
```

```python
model=Sequential([Embedding(vocab_size,128,input_length=max_len),LSTM(256),Dense(vocab_size,activation='softmax')])
```

```python
model.compile(loss='sparse_categorical_crossentropy',metrics=['accuracy'],optimizer='adam')
```

```python
model.build(input_shape=(None,max_len))
```

```python
model.summary()
```
